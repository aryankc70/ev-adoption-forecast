```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

DATA_PATH = Path("../data/raw/ev_clean.parquet")
df = pd.read_parquet(DATA_PATH)
print(df.shape)
df.head()
```

```python
def plot_sales_share(df: pd.DataFrame, mode: str, countries: list[str], powertrain: str = "EV") -> None:
    subset = df[
        (df["parameter"] == "EV sales share")
        & (df["mode"] == mode)
        & (df["powertrain"] == powertrain)
        & (df["region_country"].isin(countries))
    ].sort_values("year")

    fig, ax = plt.subplots()
    sns.lineplot(data=subset, x="year", y="value", hue="region_country", marker="o", ax=ax)
    ax.set_title(f"{mode} — EV Sales Share (%) Over Time")
    ax.set_ylabel("EV Sales Share (%)")
    ax.set_xlabel("Year")
    plt.legend(title="Country", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

benchmark_countries = ["Norway", "China", "Germany", "USA", "United Kingdom", "India"]
plot_sales_share(df, mode="Cars", countries=benchmark_countries)
```

```python
benchmark_countries = ["Norway", "China", "Germany", "USA", "United Kingdom", "India", "Nepal"]
plot_sales_share(df, mode="Cars", countries=benchmark_countries)
```

```python
plot_sales_share(df, mode="2 and 3 wheelers", countries=["China", "India", "Viet Nam", "Thailand", "Indonesia", "Nepal"])
```

```python
coverage = (
    df[df["parameter"] == "EV sales share"]
    .groupby(["mode"])["region_country"]
    .nunique()
)
print("Countries with EV sales share data, by mode:")
print(coverage)
print()

cars_countries = set(df[(df["parameter"] == "EV sales share") & (df["mode"] == "Cars")]["region_country"])
wheelers_countries = set(df[(df["parameter"] == "EV sales share") & (df["mode"] == "2 and 3 wheelers")]["region_country"])

print(f"Countries with Cars data only (no 2/3-wheeler data): {len(cars_countries - wheelers_countries)}")
print(sorted(cars_countries - wheelers_countries))
print()
print(f"Countries with 2/3-wheeler data only (no Cars data): {len(wheelers_countries - cars_countries)}")
print(sorted(wheelers_countries - cars_countries))
print()
print(f"Countries with BOTH: {len(cars_countries & wheelers_countries)}")
```
